### Clasificación

```python
#Modelo simple
```

```python
X_CL2, y_CL2 = make_blobs(n_samples = 100, n_features = 2, centers = 8, cluster_std = 1.3,random_state = 4)
y_CL2 = y_CL2 % 2

plt.figure()
plt.title('DATOS ARTIFICIALES PARA CLASIFICACION EN CLUSTERS')
plt.scatter(X_CL2[:,0], X_CL2[:,1], c=y_CL2, marker= 'o', s=50, cmap=cmap)
plt.show()
```

```python
cmap = ListedColormap(["blue", "green", "yellow", "red"])
from sklearn.neural_network import MLPClassifier
#(MLP significa Multi Layer Perceptron)
```

```python
X_test, X_train, y_test, y_train = train_test_split(X_CL2, y_CL2, random_state=0)

clasigRN1 = MLPClassifier().fit(X_test, y_test)
clasigRN1.score(X_train, y_train)
```

```python
clasigRN1.score(X_test, y_test)
```

Regularización con Alpha

### Redes Neuronales para regresión

```python
from sklearn.datasets import make_regression
plt.figure()
plt.title('Regresión simple con una variables dependiente')
X_R1, y_R1 = make_regression(n_samples = 100, n_features=1,
                            n_informative=1, bias = 150.0,
                            noise = 30, random_state=0)
plt.scatter(X_R1, y_R1, marker= 'o', s=50)
plt.show()
```

```python
from sklearn.neural_network import MLPRegressor

fig, subaxes = plt.subplots(2, 3, figsize=(11,8), dpi=70)

X_pred = np.linspace(-3, 3, 30).reshape(-1,1)

X_train, X_test, y_train, y_test = train_test_split(X_R1[0::5], y_R1[0::5], random_state = 0)

for axis_f_i, activacion_i in zip(subaxes, ['tanh', 'relu']):
    for añfa_i, axis_c_i in zip([0.0001, 1.0, 100], axis_f_i):
        RegRN = MLPRegressor(hidden_layer_sizes = [100,100], activation = activacion_i,
                             alpha = añfa_i, solver = 'lbfgs').fit(X_train, y_train)
        y_pred = RegRN.predict(X_pred)
        axis_c_i.set_xlim([-2.5, 0.75])
        axis_c_i.plot(X_pred, y_pred, '^', markersize = 10)
        axis_c_i.plot(X_train, y_train, 'o')
        axis_c_i.set_xlabel('Dependiente')
        axis_c_i.set_ylabel('Objetivo')
        axis_c_i.set_title('RN Multicapa \nalpha={}, Func act={}, Score={:.3f}'.format(añfa_i, activacion_i, RegRN.score(X_test, y_test) ))
        plt.tight_layout()
```

#### Ejemplo real, Calidad del Vino

```python
base = pd.read_csv('https://archive.ics.uci.edu/ml/machine-learning-databases/wine-quality/winequality-red.csv', delimiter=';')
```

```python
base.head()
```

```python
base['quality'].value_counts()
```

```python
base1 = base[(base['quality']==5) | (base['quality']==6)]
X = base1.drop(['quality'], axis=1)
y = base1['quality']
```

```python
from sklearn.neural_network import MLPClassifier
from sklearn.preprocessing import MinMaxScaler

normal = MinMaxScaler()

X_train, X_test, y_train, y_test = train_test_split(X, y, random_state = 0)
X_train_norm = normal.fit_transform(X_train)
X_test_norm = normal.transform(X_test)

clasifVR = MLPClassifier(hidden_layer_sizes = [100, 100, 100, 100], alpha = 1.0, activation='tanh', random_state = 0, solver='lbfgs').fit(X_train_norm, y_train)
```

```python
clasifVR.get_params()
```

```python
print('Calidad del Vino Rojo')
print('Exactitud entrenamiento: {:.3f}'.format(clasifVR.score(X_train_norm, y_train)))
print('Exactitud comprobacion: {:.3f}'.format(clasifVR.score(X_test_norm, y_test)))
```
