```python
import numpy as np
import matplotlib.pyplot as plt
from tensorflow import keras
from keras.utils import to_categorical
from keras.models import Sequential
from keras.layers import Dense, Dropout, Flatten, InputLayer
from tensorflow.keras.optimizers import Adam
```

```python
dataset = keras.datasets.mnist
(X_train, y_train), (X_test, y_test) = dataset.load_data()
```

```python
print(f'Tamaño entrenamiento: {X_train.shape}, Tamaño comprobación: {X_test.shape}')
```

```python
_, ax = plt.subplots(2, 2, figsize=(5, 5))
index = 0

for i in range(2):
    for j in range(2):
        cell = ax[i][j]
        cell.set_xticks([])
        cell.set_yticks([])
        cell.grid(False)
        cell.imshow(X_test[index], cmap='gray')
        index += 1

plt.show()
```

```python
#Normalización (MinMaxScaler)
X_train_scaled = X_train / 255.0 # Los minimos son cero (x - o)/ (Xmax - Xmin)
X_test_scaled = X_test / 255.0
```

```python
y_train[1:20]
```

```python
#Configuración del modelo
def create_model(neurons=128, dropout_rate=0.3, learning_rate=0.001):
    """
    neurons = neuronas para la capa oculta
    dropout_rate =  regularizacion, ratio de neuronas a apagarse
    learning_rate =  ratio de aprendizaje, se aconseja no moverlo
    """
    #Creación del modelo de tensorflow
    model = Sequential()
    
    #Capa de entrada 
    model.add(InputLayer(shape=(28, 28)))
    model.add(Flatten()) #Para vectorizar la imagen
    #model.add(Flatten(input_shape=(28, 28))) es una alternativa
    
    #primera capa oculta
    model.add(Dense(neurons, activation='relu')) 
    model.add(Dropout(dropout_rate)) #Regularización
    
    
    #Segunda capa oculta
    model.add(Dense(neurons, activation='relu'))
    
    #Capa de salida
    model.add(Dense(10, activation='softmax')) #se reduce a 10 neuronas, debe usarse softmax

    #Configuraciónn del optimizer y compilación del modelo
    optimizer = Adam(learning_rate= learning_rate) #Cuidado con este parámetro
    model.compile(loss='sparse_categorical_crossentropy', optimizer=optimizer, metrics=['accuracy']) #como se trata de una salida multinivel se usa sparse_categorical_crossentropy
    
    return model
```

```python
#Crear y entrenar el modelo GridSearch
model = create_model()
history = model.fit(X_train_scaled, y_train, epochs=10, batch_size=32, validation_data=(X_test_scaled, y_test), verbose=1)
```

```python
#Historial de funciºon de perdida
losses = history.history['loss']
val_losses = history.history['val_loss']
n = len(losses)
iters = np.arange(n) + 1
plt.figure(figsize=(5, 5))
plt.plot(iters, losses, label='Entrenamiento')
plt.plot(iters, val_losses, label='Validación')
plt.xlabel('Iteraciones')
plt.ylabel('Loss')
plt.title('Evolucion del loss')
plt.legend()
```
