### Redes preentrenadas VGG-16

![vgg16](vgg162.jpg)

```python
from tensorflow.keras import Sequential, models
from tensorflow.keras.layers import Dense, Flatten
from tensorflow.keras.applications import VGG16
from tensorflow.keras.optimizers import Adam
from tensorflow.keras.datasets import cifar10
from tensorflow.keras.utils import to_categorical
```

```python
#Cargar el conjunto de datos CIFAR-10
#(train_images, train_labels), (test_images, test_labels) = cifar10.load_data()

# Normalizar las imágenes al rango [0, 1]
#train_images = train_images / 255.0
#test_images = test_images / 255.0
```

```python
# Convertir las etiquetas a one-hot encoding
train_labels = to_categorical(train_labels, 10)
test_labels = to_categorical(test_labels, 10)
```

```python
# Cargar VGG16 preentrenada (excluyendo las capas fully connected)
base_model = VGG16(include_top=False, input_shape=(32, 32, 3)) #weights='imagenet',
```

```python
# Congelar las capas convolucionales de VGG16
for layer in base_model.layers:
    layer.trainable = False

#for layer in base_model.layers[-4:]:  #si queremos congelar solo las ultimas capas
#    layer.trainable = True
```

```python
#Crear el modelo
model = Sequential()
#Añadir la base de VGG16
model.add(base_model)
```

```python
# Añadir nuevas capas fully connected
model.add(Flatten())
model.add(Dense(128, activation='relu'))
model.add(Dense(10, activation='softmax'))  # 10 clases para CIFAR-10
```

```python
# Compilar el modelo
model.compile(optimizer=Adam(learning_rate=0.001),
              loss='categorical_crossentropy',
              metrics=['accuracy'])
# Resumen del modelo
model.summary()
```

```python
#Entrenar el modelo
history = model.fit(train_images, train_labels, epochs=5, #batch_size=64,
                    validation_data=(test_images, test_labels))
```

```python
# Evaluar el modelo
test_loss, test_acc = model.evaluate(test_images, test_labels, verbose=2)
print(f"\nPrecisión en el conjunto de prueba: {test_acc:.4f}")
```

#### Mejora del entrenamiento

```python
#crear modificaciones del set original

from tensorflow.keras.preprocessing.image import ImageDataGenerator

train_datagen = ImageDataGenerator(
    rotation_range=15,
    width_shift_range=0.1,
    height_shift_range=0.1,
    horizontal_flip=True,
    rescale=1./255
)

train_generator = train_datagen.flow(train_images, train_labels, batch_size=64)
```

```python
#Añadiremos un callback para guardar el mejor modelo y la historia
from tensorflow.keras.callbacks import ModelCheckpoint

checkpoint_callback = ModelCheckpoint(
    filepath='best_model.keras',
    monitor='val_accuracy',
    mode='max',
    save_best_only=True
)
```

```python
history = model.fit(train_images, train_labels, epochs=5, #batch_size=64,
                    validation_data=(test_images, test_labels),
                    callbacks=[checkpoint_callback])
```

```python
# Evaluar el modelo
test_loss, test_acc = model.evaluate(test_images, test_labels, verbose=2)
print(f"\nPrecisión en el conjunto de prueba: {test_acc:.4f}")
```

```python
# Cargar el mejor modelo guardado
from tensorflow.keras.models import load_model
best_model2 = load_model('best_model.keras')

# Evaluar el mejor modelo
test_loss, test_acc = best_model2.evaluate(test_images, test_labels, verbose=2)
print(f"Precisión del mejor modelo en el conjunto de prueba: {test_acc:.4f}")
```
